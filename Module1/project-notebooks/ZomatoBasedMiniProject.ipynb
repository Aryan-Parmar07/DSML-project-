# Mini Project: Zomato Restaurant Data Exploration
Perform end-to-end Exploratory Data Analysis (EDA) on restaurant listing data to uncover patterns in ratings, dining costs, price tiers, and popularity.

### Pipeline:
1. **Data Ingestion & Inspection**
2. **Data Cleaning & Missing Value Handling**
3. **Exploratory Visualizations**
4. **Key Insights & Findings**

```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

# Load dataset with local/Colab environment fallback
data_path = "../datasets/zomato_eda_mini_project.csv" if os.path.exists("../datasets/zomato_eda_mini_project.csv") else "/content/zomato_eda_mini_project.csv"
df = pd.read_csv(data_path)

print(f"Dataset Shape: {df.shape}")
df.head()
```

## 1. Data Inspection & Cleaning

```python
# Dataset info and column types
df.info()
```

```python
# Check missing values before cleaning
print("Missing values before cleaning:")
display(df.isnull().sum())

# Check and remove duplicates
print(f"\nDuplicate rows found: {df.duplicated().sum()}")
df.drop_duplicates(inplace=True)
print(f"Rows remaining after deduplication: {len(df)}")
```

```python
# Impute missing values and prepare clean columns
df["Cuisines"] = df["Cuisines"].fillna("Unknown")
df["Average Cost for two"] = df["Average Cost for two"].fillna(df["Average Cost for two"].median())
df["Rating"] = df["Aggregate rating"].astype(float)

print("Missing values after cleaning:")
display(df.isnull().sum())
```

## 2. Exploratory Data Visualizations

### Chart 1: Distribution of Restaurants Across Price Ranges

```python
plt.figure(figsize=(8, 4.5))
palette = sns.color_palette("mako", len(df["Price range"].unique()))
ax = sns.countplot(data=df, x="Price range", palette=palette, hue="Price range", legend=False)

plt.title("Number of Restaurants by Price Range", fontsize=12, fontweight="bold")
plt.xlabel("Price Range (1 = Budget, 4 = Fine Dining)", fontsize=10)
plt.ylabel("Number of Restaurants", fontsize=10)

for p in ax.patches:
    ax.annotate(f"{int(p.get_height())}", (p.get_x() + p.get_width() / 2., p.get_height()),
                ha='center', va='baseline', fontsize=9, xytext=(0, 4), textcoords='offset points')

plt.tight_layout()
plt.show()
```

### Chart 2: Average Customer Rating by Price Range

```python
plt.figure(figsize=(8, 4.5))
sns.barplot(data=df, x="Price range", y="Rating", palette="Blues_d", hue="Price range", legend=False, errorbar=None)

plt.title("Average Restaurant Rating by Price Range", fontsize=12, fontweight="bold")
plt.xlabel("Price Range", fontsize=10)
plt.ylabel("Average Rating (out of 5)", fontsize=10)
plt.ylim(0, 5)

plt.tight_layout()
plt.show()
```

### Chart 3: Distribution of Customer Ratings

```python
plt.figure(figsize=(8, 4.5))
sns.histplot(df["Rating"], bins=12, kde=True, color="#2563eb")

mean_rating = df["Rating"].mean()
plt.axvline(mean_rating, color="#dc2626", linestyle="--", label=f"Mean Rating: {mean_rating:.2f}")

plt.title("Overall Rating Distribution", fontsize=12, fontweight="bold")
plt.xlabel("Rating", fontsize=10)
plt.ylabel("Frequency", fontsize=10)
plt.legend()
plt.tight_layout()
plt.show()
```

### Chart 4: Average Cost for Two vs. Rating

```python
plt.figure(figsize=(9, 5))
sns.scatterplot(
    data=df,
    x="Average Cost for two",
    y="Rating",
    hue="Price range",
    palette="viridis",
    size="Votes",
    sizes=(30, 200),
    alpha=0.85
)

plt.title("Dining Cost for Two vs. Customer Rating", fontsize=12, fontweight="bold")
plt.xlabel("Average Cost for Two (₹)", fontsize=10)
plt.ylabel("Rating", fontsize=10)
plt.tight_layout()
plt.show()
```

## 3. Summary & Key Findings
- **Price Range vs. Rating**: Higher price range restaurants generally enjoy higher average ratings.
- **Budget Dominance**: The vast majority of listed restaurants fall into lower price range tiers (1 and 2).
- **Data Quality**: Imputing missing cuisines with `'Unknown'` and median filling for cost ensures no data loss during analysis.
