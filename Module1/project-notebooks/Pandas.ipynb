```python
import os
import pandas as pd
import numpy as np
```

## 1. Creating and Inspecting a DataFrame

```python
# Create a sample DataFrame
data = {
    "Name": ["Asif", "Aman", "Arun", "Ajay", "Alex", "Appu", "Pappu"],
    "Age":  [32, 33, 23, 24, 35, 26, 18],
    "City": ["Del", "Blr", "Mlr", "Kol", "Ker", "Mys", "Chn"]
}

df = pd.DataFrame(data)
df
```

### Basic Inspections (`head`, `tail`, `shape`, `info`, `describe`)

```python
print(f"DataFrame Shape (rows, columns): {df.shape}")
print(f"Columns: {list(df.columns)}")
df.info()
```

```python
# Statistical summary of numerical columns
df.describe()
```

## 2. Row & Column Selection (`loc`, `iloc`, boolean indexing)

```python
# Row-wise selection: label-based (.loc) vs integer position-based (.iloc)
print("--- Using .loc (rows 3 to 4 inclusive) ---")
display(df.loc[3:4])

print("--- Using .iloc (row index 3) ---")
display(df.iloc[[3]])

# Column-wise selection
display(df[["Name", "City"]].head(3))
```

## 3. Adding, Modifying, and Filtering Columns

```python
# Add a new column 'Salary'
df["Salary"] = [20000, 340000, 43000, 23000, 450000, 56000, 240000]

# Modify an existing column
df["Age"] = df["Age"] + 1

# Filter: Employees older than 25 with salary > 50,000
high_earners = df[(df["Age"] > 25) & (df["Salary"] > 50000)]

print("Updated DataFrame:")
display(df)
print("\nFiltered High Earners (Age > 25 & Salary > 50k):")
display(high_earners)
```

## 4. GroupBy & Multi-Metric Aggregations

```python
# Group by city and compute multiple metrics
city_summary = df.groupby("City").agg(
    Avg_Salary=("Salary", "mean"),
    Max_Salary=("Salary", "max"),
    Total_Salary=("Salary", "sum"),
    Employee_Count=("Age", "count")
).reset_index()

city_summary
```
